````python
import json
import re

def count_sentences(text):
    """Число предложений в тексте. Точка внутри числа концом предложения не считается."""
    return len([s for s in re.split(r"[.!?]+(?!\d)", text) if s.strip()])


REQUIRED_FIELDS = ("product_id", "description", "pros", "cons", "tags")
LIST_LIMITS = {"pros": (2, 5), "cons": (1, 3), "tags": (3, 8)}


def validate_card(payload, product_id):
    """Вернуть список нарушений. Пустой список — карточка валидна."""
    if not isinstance(payload, dict):
        return ["Ответ не является JSON-объектом"]

    errors = []
    missing = [field for field in REQUIRED_FIELDS if field not in payload]
    if missing:
        errors.append(f"Нет полей: {', '.join(missing)}")

    if "product_id" in payload and payload["product_id"] != product_id:
        errors.append(
            f"product_id {payload['product_id']!r} не совпадает с {product_id!r}"
        )

    if "description" in payload:
        description = payload["description"]
        if not isinstance(description, str):
            errors.append("description должен быть строкой")
        else:
            length = len(description)
            if not 120 <= length <= 700:
                errors.append(f"description: {length} символов, нужно 120–700")
            sentences = count_sentences(description)
            if not 2 <= sentences <= 5:
                errors.append(f"description: {sentences} предложений, нужно 2–5")

    for name, (low, high) in LIST_LIMITS.items():
        if name not in payload:
            continue
        value = payload[name]
        if (
            not isinstance(value, list)
            or not all(isinstance(item, str) and item.strip() for item in value)
        ):
            errors.append(f"{name}: нужен список непустых строк")
            continue
        if not low <= len(value) <= high:
            errors.append(f"{name}: {len(value)} пунктов, нужно {low}–{high}")

    return errors


from transformers import AutoTokenizer

MAX_INPUT_TOKENS = 1500
_REVIEW_LEAD_INS = (
    "что могу сказать:",
    "коротко:",
    "в целом",
    "пользуюсь пару недель.",
    "пользуюсь месяц.",
    "пользуюсь каждый день.",
    "купил на замену старому.",
    "брал для себя.",
    "заказывал на подарок.",
)


def review_core(text):
    normalized = " ".join(text.lower().split())
    changed = True
    while changed:
        changed = False
        for lead in _REVIEW_LEAD_INS:
            if normalized.startswith(lead):
                normalized = normalized[len(lead):].strip(" :.-")
                changed = True
                break
    return normalized


def unique_reviews(reviews):
    """Один отзыв на смысл. Из дублей остаётся самая короткая формулировка."""
    best = {}
    order = []
    for review in reviews:
        text = str(review).strip()
        if not text:
            continue
        key = review_core(text)
        if not key:
            continue
        current = best.get(key)
        if current is None:
            best[key] = text
            order.append(key)
        elif len(text) < len(current):
            best[key] = text
    return [best[key] for key in order]


def spread_reviews(items, count):
    """Равномерно взять count отзывов, включая первый и последний."""
    if count >= len(items):
        return list(items)
    if count <= 0:
        return []
    if count == 1:
        return [items[0]]
    last = len(items) - 1
    picked = []
    seen = set()
    for index in range(count):
        position = round(index * last / (count - 1))
        if position not in seen:
            seen.add(position)
            picked.append(items[position])
    return picked


def fit_product(card, token_count, max_tokens=MAX_INPUT_TOKENS):
    """Оставить уникальные отзывы и уложиться в max_tokens на весь промпт."""
    def count_reviews(reviews):
        return token_count({**card, "reviews": reviews})

    unique = unique_reviews(card.get("reviews") or [])
    if not unique or count_reviews(unique) <= max_tokens:
        selected = unique
    else:
        selected = []
        low, high = 0, len(unique)
        while low <= high:
            mid = (low + high) // 2
            candidate = spread_reviews(unique, mid)
            if count_reviews(candidate) <= max_tokens:
                selected = candidate
                low = mid + 1
            else:
                high = mid - 1
    fitted = dict(card)
    fitted["reviews"] = selected
    return fitted


tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")


def count_prompt_tokens(product):
    """Токены всего запроса так, как их видит Qwen."""
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": json.dumps(product, ensure_ascii=False)},
    ]
    encoded = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_dict=False,
    )
    return len(encoded)


def loads_model_json(content):
    """Достать JSON-объект, даже если модель обернула его в ```json."""
    text = content.strip()
    if text.startswith("```"):
        lines = text.splitlines()
        lines = lines[1:]
        if lines and lines[-1].strip().startswith("```"):
            lines = lines[:-1]
        text = "\n".join(lines).strip()
    start = text.find("{")
    end = text.rfind("}")
    if start != -1 and end > start:
        text = text[start:end + 1]
    return json.loads(text)


results = []
for index, product in enumerate(data, start=1):
    product_id = product["product_id"]
    full_tokens = count_prompt_tokens(product)
    original_reviews = len(product.get("reviews") or [])
    product = fit_product(product, count_prompt_tokens)
    fitted_tokens = count_prompt_tokens(product)
    response = client.chat.completions.create(
        model="Qwen/Qwen2.5-1.5B-Instruct",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": json.dumps(product, ensure_ascii=False)},
        ],
        temperature=0.2,
        extra_body={"structured_outputs": {"json": json_schema}},
    )
    content = (response.choices[0].message.content or "").strip()
    try:
        payload = loads_model_json(content)
    except json.JSONDecodeError as error:
        preview = content[:120].replace("\n", " ") or "пустой ответ"
        errors = [f"Не JSON: {error.msg}. Начало ответа: {preview!r}"]
        payload = None
    else:
        errors = validate_card(payload, product_id)

    valid = not errors
    print(
        f"[{index}/{len(data)}] {product_id}: "
        f"отзывов {original_reviews}->{len(product['reviews'])}, "
        f"токенов {full_tokens}->{fitted_tokens}, "
        f"{'валидна' if valid else 'невалидна'}"
    )
    for error in errors:
        print(f"  - {error}")
    results.append(
        {"product_id": product_id, "valid": valid, "errors": errors, "card": payload}
    )

valid_count = sum(item["valid"] for item in results)
print(f"\nВалидных карточек: {valid_count} из {len(results)}")
````

````python
CARD_LIMIT = 20
BASELINE = {
    "temperature": 0.3,
    "top_p": 0.9,
    "top_k": 50,
    "repetition_penalty": 1.1,
    "max_tokens": 350,
}
SWEEP = {
    "temperature": [0.0, 0.3, 0.7],
    "top_p": [0.9, 1.0],
    "top_k": [20, 50, None],
    "repetition_penalty": [1.0, 1.1, 1.2],
    "max_tokens": [250, 350, 500],
}
PARAMETER_ORDER = list(SWEEP)


def format_value(parameter, value):
    if parameter == "top_k" and value is None:
        return "без ограничения"
    if isinstance(value, float):
        return f"{value:.1f}"
    return str(value)


def has_overlap(items):
    normalized = [item.lower().strip() for item in items]
    for index, left in enumerate(normalized):
        for right in normalized[index + 1:]:
            if left and right and (left == right or left in right or right in left):
                return True
    return False


def build_runs():
    """Одна комбинация на каждое значение. База считается один раз."""
    runs = []
    seen = {}
    for parameter, values in SWEEP.items():
        for value in values:
            config = dict(BASELINE)
            config[parameter] = value
            key = tuple(config[name] for name in PARAMETER_ORDER)
            run = seen.get(key)
            if run is None:
                run = {"config": config, "rows": []}
                seen[key] = run
                runs.append(run)
            run["rows"].append({"parameter": parameter, "value": value})
    return runs


def mean(values):
    return sum(values) / len(values) if values else None


def format_number(value):
    if value is None:
        return "—"
    return f"{value:.1f}"


def evaluate_card(card, config):
    result = {
        "product_id": card["product_id"],
        "valid": False,
        "description_length": None,
        "completion_tokens": None,
        "truncated": False,
        "tag_overlap": False,
        "error": None,
    }
    extra_body = {"structured_outputs": {"json": json_schema}}
    if config["top_k"] is not None:
        extra_body["top_k"] = config["top_k"]
    extra_body["repetition_penalty"] = config["repetition_penalty"]
    try:
        response = client.chat.completions.create(
            model="Qwen/Qwen2.5-1.5B-Instruct",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": json.dumps(card, ensure_ascii=False)},
            ],
            temperature=config["temperature"],
            top_p=config["top_p"],
            max_tokens=config["max_tokens"],
            extra_body=extra_body,
        )
    except Exception as error:
        result["error"] = str(error)
        return result

    choice = response.choices[0]
    usage = response.usage
    result["completion_tokens"] = usage.completion_tokens if usage is not None else None
    result["truncated"] = choice.finish_reason == "length"
    content = (choice.message.content or "").strip()
    try:
        payload = loads_model_json(content)
    except json.JSONDecodeError as error:
        result["error"] = str(error)
        return result
    errors = validate_card(payload, card["product_id"])
    if errors:
        result["error"] = "; ".join(errors)
        return result
    result["valid"] = True
    result["description_length"] = len(payload["description"])
    result["tag_overlap"] = has_overlap(payload["pros"]) or has_overlap(payload["tags"])
    return result


def summarize(parameter, value, outcomes):
    valid = [item for item in outcomes if item["valid"]]
    tokens = [item["completion_tokens"] for item in outcomes if item["completion_tokens"] is not None]
    return {
        "parameter": parameter,
        "value": value,
        "value_label": format_value(parameter, value),
        "total": len(outcomes),
        "valid_count": len(valid),
        "valid_rate": len(valid) / len(outcomes) if outcomes else 0.0,
        "avg_description_length": mean([item["description_length"] for item in valid]),
        "avg_completion_tokens": mean(tokens),
        "truncated": sum(1 for item in outcomes if item["truncated"]),
        "tag_overlap_count": sum(1 for item in valid if item["tag_overlap"]),
    }


def choose_by_rate(rows, preferred):
    best_rate = max(row["valid_rate"] for row in rows)
    contenders = [row for row in rows if row["valid_rate"] == best_rate]
    for row in contenders:
        if row["value"] == preferred:
            return row
    return contenders[0]


def choose_penalty(rows):
    best_rate = max(row["valid_rate"] for row in rows)
    contenders = [row for row in rows if row["valid_rate"] == best_rate]
    return min(
        contenders,
        key=lambda row: (
            row["tag_overlap_count"],
            0 if row["value"] == BASELINE["repetition_penalty"] else 1,
        ),
    )


def choose_max_tokens(rows):
    without_cutoff = [row for row in rows if row["truncated"] == 0]
    if without_cutoff:
        return min(without_cutoff, key=lambda row: row["value"])
    return max(rows, key=lambda row: (row["valid_rate"], row["value"]))


def print_table(table):
    header = (
        f"{'\u043f\u0430\u0440\u0430\u043c\u0435\u0442\u0440':<22} {'\u0437\u043d\u0430\u0447\u0435\u043d\u0438\u0435':<18} {'\u0432\u0430\u043b\u0438\u0434\u043d\u044b\u0435':>10} "
        f"{'доля':>8} {'длина':>8} {'токены':>8} {'обрывы':>8} {'повторы':>8}"
    )
    print(header)
    print("-" * len(header))
    for row in table:
        print(
            f"{row['parameter']:<22} {row['value_label']:<18} "
            f"{row['valid_count']}/{row['total']:>2} "
            f"{row['valid_rate']:>7.0%} "
            f"{format_number(row['avg_description_length']):>8} "
            f"{format_number(row['avg_completion_tokens']):>8} "
            f"{row['truncated']:>8} {row['tag_overlap_count']:>8}"
        )


cards = [fit_product(card, count_prompt_tokens) for card in data[:CARD_LIMIT]]
runs = build_runs()
print(
    f"Карточек: {len(cards)}. Комбинаций: {len(runs)}. "
    "В каждой от базы отличается один параметр."
)
print(
    "База: temperature 0.3, top_p 0.9, top_k 50, "
    "repetition_penalty 1.1, max_tokens 350."
)

sweep_table = []
sweep_runs = []
for run_index, run in enumerate(runs, start=1):
    config = run["config"]
    labels = ", ".join(
        f"{row['parameter']}={format_value(row['parameter'], row['value'])}"
        for row in run["rows"]
    )
    print(f"\n[{run_index}/{len(runs)}] {labels}", flush=True)
    outcomes = []
    for index, card in enumerate(cards, start=1):
        outcome = evaluate_card(card, config)
        outcomes.append(outcome)
        mark = "ok" if outcome["valid"] else "fail"
        print(f"  {index:02}/{len(cards)} {card['product_id']} {mark}", flush=True)
    valid_count = sum(item["valid"] for item in outcomes)
    print(f"  валидных {valid_count}/{len(cards)}")
    sweep_runs.append({"config": config, "outcomes": outcomes})
    for row in run["rows"]:
        sweep_table.append(summarize(row["parameter"], row["value"], outcomes))

print("\nПараметр — значение — доля валидных — средняя длина описания — выходные токены")
print_table(sweep_table)

by_parameter = {
    parameter: [row for row in sweep_table if row["parameter"] == parameter]
    for parameter in PARAMETER_ORDER
}
chosen_rows = {
    "temperature": choose_by_rate(by_parameter["temperature"], BASELINE["temperature"]),
    "top_p": choose_by_rate(by_parameter["top_p"], BASELINE["top_p"]),
    "top_k": choose_by_rate(by_parameter["top_k"], BASELINE["top_k"]),
    "repetition_penalty": choose_penalty(by_parameter["repetition_penalty"]),
    "max_tokens": choose_max_tokens(by_parameter["max_tokens"]),
}
chosen = {parameter: row["value"] for parameter, row in chosen_rows.items()}

print("\nВыбранный набор")
for parameter in PARAMETER_ORDER:
    row = chosen_rows[parameter]
    peers = by_parameter[parameter]
    peer_text = "; ".join(
        f"{item['value_label']}: {item['valid_count']}/{item['total']}, "
        f"длина {format_number(item['avg_description_length'])}, "
        f"токены {format_number(item['avg_completion_tokens'])}, "
        f"обрывов {item['truncated']}, повторов {item['tag_overlap_count']}"
        for item in peers
    )
    if parameter == "max_tokens" and row["truncated"] == 0:
        reason = "наименьшее значение без обрывов JSON"
    elif parameter == "max_tokens":
        reason = "обрывы есть на всех значениях, взято лучшее по доле валидных"
    elif parameter == "repetition_penalty":
        reason = "при лучшей доле валидных меньше повторов в pros и tags"
    else:
        reason = "выше доля валидных; при равенстве остаётся база"
    print(f"{parameter}={row['value_label']}: {reason}.")
    print(f"  {peer_text}")

print("\nchosen =", {name: format_value(name, value) for name, value in chosen.items()})
````

````python
CHOSEN = {
    "temperature": 0.0,
    "top_p": 1.0,
    "top_k": 20,
    "repetition_penalty": 1.2,
    "max_tokens": 500,
}


def same_config(left, right):
    return all(left[name] == right[name] for name in BASELINE)


def run_config(config, label):
    print(f"\n{label}", flush=True)
    outcomes = []
    for index, card in enumerate(cards, start=1):
        outcome = evaluate_card(card, config)
        outcomes.append(outcome)
        mark = "ok" if outcome["valid"] else "fail"
        print(f"  {index:02}/{len(cards)} {card['product_id']} {mark}", flush=True)
    valid_count = sum(item["valid"] for item in outcomes)
    print(f"  валидных {valid_count}/{len(cards)}")
    return outcomes


def pack(outcomes):
    valid = [item for item in outcomes if item["valid"]]
    tokens = [item["completion_tokens"] for item in outcomes if item["completion_tokens"] is not None]
    return {
        "valid_count": len(valid),
        "total": len(outcomes),
        "avg_description_length": mean([item["description_length"] for item in valid]),
        "avg_completion_tokens": mean(tokens),
        "truncated": sum(1 for item in outcomes if item["truncated"]),
        "tag_overlap_count": sum(1 for item in valid if item["tag_overlap"]),
    }


baseline_outcomes = None
for run in sweep_runs:
    if same_config(run["config"], BASELINE):
        baseline_outcomes = run["outcomes"]
        break
if baseline_outcomes is None:
    baseline_outcomes = run_config(BASELINE, "База")
else:
    print("База взята из сетки, повторно не гоняется.")

chosen_outcomes = run_config(CHOSEN, "Выбранный набор")

base_pack = pack(baseline_outcomes)
chosen_pack = pack(chosen_outcomes)
rows = (("база", base_pack), ("выбранный", chosen_pack))

header = (
    f"{'набор':<12} {'валидные':>10} {'доля':>8} {'длина':>8} "
    f"{'токены':>8} {'обрывы':>8} {'повторы':>8}"
)
print("\n" + header)
print("-" * len(header))
for name, item in rows:
    print(
        f"{name:<12} {item['valid_count']}/{item['total']:<7} "
        f"{item['valid_count'] / item['total']:>7.0%} "
        f"{format_number(item['avg_description_length']):>8} "
        f"{format_number(item['avg_completion_tokens']):>8} "
        f"{item['truncated']:>8} {item['tag_overlap_count']:>8}"
    )

print("\nПо карточкам")
both = only_base = only_chosen = neither = 0
for base, chosen in zip(baseline_outcomes, chosen_outcomes):
    if base["valid"] and chosen["valid"]:
        both += 1
        mark = "оба"
    elif base["valid"]:
        only_base += 1
        mark = "только база"
    elif chosen["valid"]:
        only_chosen += 1
        mark = "только выбранный"
    else:
        neither += 1
        mark = "оба нет"
    print(f"  {base['product_id']}: {mark}")

print(
    f"\nОба валидны: {both}. Только база: {only_base}. "
    f"Только выбранный: {only_chosen}. Оба нет: {neither}."
)
````

````python
import json
import re

if "CHOSEN" not in globals():
    CHOSEN = {
        "temperature": 0.0,
        "top_p": 1.0,
        "top_k": 20,
        "repetition_penalty": 1.2,
        "max_tokens": 500,
    }

if "fit_product" not in globals():
    def count_sentences(text):
        """Число предложений в тексте. Точка внутри числа концом предложения не считается."""
        return len([s for s in re.split(r"[.!?]+(?!\d)", text) if s.strip()])


    REQUIRED_FIELDS = ("product_id", "description", "pros", "cons", "tags")
    LIST_LIMITS = {"pros": (2, 5), "cons": (1, 3), "tags": (3, 8)}


    def validate_card(payload, product_id):
        """Вернуть список нарушений. Пустой список — карточка валидна."""
        if not isinstance(payload, dict):
            return ["Ответ не является JSON-объектом"]

        errors = []
        missing = [field for field in REQUIRED_FIELDS if field not in payload]
        if missing:
            errors.append(f"Нет полей: {', '.join(missing)}")

        if "product_id" in payload and payload["product_id"] != product_id:
            errors.append(
                f"product_id {payload['product_id']!r} не совпадает с {product_id!r}"
            )

        if "description" in payload:
            description = payload["description"]
            if not isinstance(description, str):
                errors.append("description должен быть строкой")
            else:
                length = len(description)
                if not 120 <= length <= 700:
                    errors.append(f"description: {length} символов, нужно 120–700")
                sentences = count_sentences(description)
                if not 2 <= sentences <= 5:
                    errors.append(f"description: {sentences} предложений, нужно 2–5")

        for name, (low, high) in LIST_LIMITS.items():
            if name not in payload:
                continue
            value = payload[name]
            if (
                not isinstance(value, list)
                or not all(isinstance(item, str) and item.strip() for item in value)
            ):
                errors.append(f"{name}: нужен список непустых строк")
                continue
            if not low <= len(value) <= high:
                errors.append(f"{name}: {len(value)} пунктов, нужно {low}–{high}")

        return errors


    from transformers import AutoTokenizer

    MAX_INPUT_TOKENS = 1500
    _REVIEW_LEAD_INS = (
        "что могу сказать:",
        "коротко:",
        "в целом",
        "пользуюсь пару недель.",
        "пользуюсь месяц.",
        "пользуюсь каждый день.",
        "купил на замену старому.",
        "брал для себя.",
        "заказывал на подарок.",
    )


    def review_core(text):
        normalized = " ".join(text.lower().split())
        changed = True
        while changed:
            changed = False
            for lead in _REVIEW_LEAD_INS:
                if normalized.startswith(lead):
                    normalized = normalized[len(lead):].strip(" :.-")
                    changed = True
                    break
        return normalized


    def unique_reviews(reviews):
        """Один отзыв на смысл. Из дублей остаётся самая короткая формулировка."""
        best = {}
        order = []
        for review in reviews:
            text = str(review).strip()
            if not text:
                continue
            key = review_core(text)
            if not key:
                continue
            current = best.get(key)
            if current is None:
                best[key] = text
                order.append(key)
            elif len(text) < len(current):
                best[key] = text
        return [best[key] for key in order]


    def spread_reviews(items, count):
        """Равномерно взять count отзывов, включая первый и последний."""
        if count >= len(items):
            return list(items)
        if count <= 0:
            return []
        if count == 1:
            return [items[0]]
        last = len(items) - 1
        picked = []
        seen = set()
        for index in range(count):
            position = round(index * last / (count - 1))
            if position not in seen:
                seen.add(position)
                picked.append(items[position])
        return picked


    def fit_product(card, token_count, max_tokens=MAX_INPUT_TOKENS):
        """Оставить уникальные отзывы и уложиться в max_tokens на весь промпт."""
        def count_reviews(reviews):
            return token_count({**card, "reviews": reviews})

        unique = unique_reviews(card.get("reviews") or [])
        if not unique or count_reviews(unique) <= max_tokens:
            selected = unique
        else:
            selected = []
            low, high = 0, len(unique)
            while low <= high:
                mid = (low + high) // 2
                candidate = spread_reviews(unique, mid)
                if count_reviews(candidate) <= max_tokens:
                    selected = candidate
                    low = mid + 1
                else:
                    high = mid - 1
        fitted = dict(card)
        fitted["reviews"] = selected
        return fitted


    tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")


    def count_prompt_tokens(product):
        """Токены всего запроса так, как их видит Qwen."""
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": json.dumps(product, ensure_ascii=False)},
        ]
        encoded = tokenizer.apply_chat_template(
            messages,
            tokenize=True,
            add_generation_prompt=True,
            return_dict=False,
        )
        return len(encoded)


    def loads_model_json(content):
        """Достать JSON-объект, даже если модель обернула его в ```json."""
        text = content.strip()
        if text.startswith("```"):
            lines = text.splitlines()
            lines = lines[1:]
            if lines and lines[-1].strip().startswith("```"):
                lines = lines[:-1]
            text = "\n".join(lines).strip()
        start = text.find("{")
        end = text.rfind("}")
        if start != -1 and end > start:
            text = text[start:end + 1]
        return json.loads(text)

import asyncio
import time

from openai import AsyncOpenAI

CONCURRENCY = 4
async_client = AsyncOpenAI(base_url="http://localhost:8000/v1", api_key="local-vllm")


async def run_concurrent(items, concurrency, generate_one):
    """Отправляет запросы параллельно, не больше `concurrency` одновременно.

    generate_one — асинхронная функция, принимает один товар и возвращает
    словарь с результатом. Возвращает список результатов и общее время прогона.
    """
    sem = asyncio.Semaphore(concurrency)

    async def wrapped(item):
        async with sem:
            return await generate_one(item)

    started = time.perf_counter()
    results = await asyncio.gather(*[wrapped(it) for it in items])
    return results, time.perf_counter() - started


async def generate_one(card):
    started = time.perf_counter()
    result = {
        "product_id": card["product_id"],
        "valid": False,
        "description_length": None,
        "completion_tokens": None,
        "latency": None,
        "error": None,
    }
    extra_body = {
        "structured_outputs": {"json": json_schema},
        "repetition_penalty": CHOSEN["repetition_penalty"],
    }
    if CHOSEN["top_k"] is not None:
        extra_body["top_k"] = CHOSEN["top_k"]
    try:
        response = await async_client.chat.completions.create(
            model="Qwen/Qwen2.5-1.5B-Instruct",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": json.dumps(card, ensure_ascii=False)},
            ],
            temperature=CHOSEN["temperature"],
            top_p=CHOSEN["top_p"],
            max_tokens=CHOSEN["max_tokens"],
            extra_body=extra_body,
        )
    except Exception as error:
        result["error"] = str(error)
        result["latency"] = time.perf_counter() - started
        print(f"  {card['product_id']} fail", flush=True)
        return result

    choice = response.choices[0]
    usage = response.usage
    result["completion_tokens"] = usage.completion_tokens if usage is not None else None
    result["latency"] = time.perf_counter() - started
    content = (choice.message.content or "").strip()
    try:
        payload = loads_model_json(content)
    except json.JSONDecodeError as error:
        result["error"] = str(error)
        print(f"  {card['product_id']} fail", flush=True)
        return result
    errors = validate_card(payload, card["product_id"])
    if errors:
        result["error"] = "; ".join(errors)
        print(f"  {card['product_id']} fail", flush=True)
        return result
    result["valid"] = True
    result["description_length"] = len(payload["description"])
    print(f"  {card['product_id']} ok", flush=True)
    return result


items = [fit_product(card, count_prompt_tokens) for card in data[:20]]
print(f"Карточек: {len(items)}, одновременно: {CONCURRENCY}")
async_results, elapsed = await run_concurrent(items, CONCURRENCY, generate_one)

valid_count = sum(item["valid"] for item in async_results)
latencies = [item["latency"] for item in async_results if item["latency"] is not None]
print(
    f"\nВалидных: {valid_count}/{len(async_results)}. "
    f"Время: {elapsed:.1f} с. "
    f"Скорость: {len(async_results) / elapsed:.2f} карточек/с. "
    f"Средняя задержка: {sum(latencies) / len(latencies):.2f} с."
)
````

````python
import json
from pathlib import Path

BENCHMARK_PATH = Path("benchmark.jsonl")
OUTPUT_PATH = Path("outputs/predictions.jsonl")
PARTIAL_PATH = Path("outputs/predictions.partial.jsonl")
CARD_FIELDS = ("product_id", "description", "pros", "cons", "tags")

benchmark = []
with BENCHMARK_PATH.open(encoding="utf-8") as source:
    for line in source:
        line = line.strip()
        if line:
            benchmark.append(json.loads(line))

items = [fit_product(card, count_prompt_tokens) for card in benchmark]
print(
    f"Карточек: {len(items)}, одновременно: {CONCURRENCY}. "
    "По прогону 20 карточек это около 15 минут."
)

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
PARTIAL_PATH.write_text("", encoding="utf-8")
write_lock = asyncio.Lock()
finished = 0


def card_record(payload, product_id):
    if not isinstance(payload, dict):
        return {"product_id": product_id}
    record = {name: payload.get(name) for name in CARD_FIELDS}
    if not record["product_id"]:
        record["product_id"] = product_id
    return record


async def generate_benchmark(card):
    global finished
    started = time.perf_counter()
    result = {
        "product_id": card["product_id"],
        "valid": False,
        "record": {"product_id": card["product_id"]},
        "completion_tokens": None,
        "latency": None,
        "error": None,
    }
    extra_body = {
        "structured_outputs": {"json": json_schema},
        "repetition_penalty": CHOSEN["repetition_penalty"],
    }
    if CHOSEN["top_k"] is not None:
        extra_body["top_k"] = CHOSEN["top_k"]
    try:
        response = await async_client.chat.completions.create(
            model="Qwen/Qwen2.5-1.5B-Instruct",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": json.dumps(card, ensure_ascii=False)},
            ],
            temperature=CHOSEN["temperature"],
            top_p=CHOSEN["top_p"],
            max_tokens=CHOSEN["max_tokens"],
            extra_body=extra_body,
            timeout=180,
        )
    except Exception as error:
        result["error"] = str(error)
        result["latency"] = time.perf_counter() - started
    else:
        choice = response.choices[0]
        usage = response.usage
        result["completion_tokens"] = usage.completion_tokens if usage is not None else None
        result["latency"] = time.perf_counter() - started
        content = (choice.message.content or "").strip()
        try:
            payload = loads_model_json(content)
        except json.JSONDecodeError as error:
            result["error"] = str(error)
        else:
            result["record"] = card_record(payload, card["product_id"])
            errors = validate_card(payload, card["product_id"])
            if errors:
                result["error"] = "; ".join(errors)
            else:
                result["valid"] = True

    async with write_lock:
        finished += 1
        with PARTIAL_PATH.open("a", encoding="utf-8") as sink:
            sink.write(json.dumps(result["record"], ensure_ascii=False) + "\n")
        if finished % 20 == 0 or finished == len(items):
            print(f"  {finished}/{len(items)}", flush=True)
        if result["error"]:
            print(f"  {card['product_id']} fail: {result['error']}", flush=True)
    return result


benchmark_results, benchmark_elapsed = await run_concurrent(
    items, CONCURRENCY, generate_benchmark
)

with OUTPUT_PATH.open("w", encoding="utf-8") as sink:
    for item in benchmark_results:
        sink.write(json.dumps(item["record"], ensure_ascii=False) + "\n")

valid_count = sum(item["valid"] for item in benchmark_results)
latencies = [item["latency"] for item in benchmark_results if item["latency"] is not None]
print(
    f"\nВалидных: {valid_count}/{len(benchmark_results)}. "
    f"Время: {benchmark_elapsed:.1f} с. "
    f"Скорость: {len(benchmark_results) / benchmark_elapsed:.2f} карточек/с. "
    f"Средняя задержка: {sum(latencies) / len(latencies):.2f} с."
)
print(f"Карточки: {OUTPUT_PATH}")
````

````python
CONCURRENCY_LEVELS = (1, 4, 16, 32)
SPEED_SUBSET = 32
BENCHMARK_LIMIT_SECONDS = 8 * 60

speed_source = []
with open("benchmark.jsonl", encoding="utf-8") as source:
    for line in source:
        line = line.strip()
        if line:
            speed_source.append(json.loads(line))
speed_items = [
    fit_product(card, count_prompt_tokens) for card in speed_source[:SPEED_SUBSET]
]
print(
    f"Один набор: {len(speed_items)} карточек. "
    f"Concurrency: {', '.join(str(level) for level in CONCURRENCY_LEVELS)}. "
    "Время на 300 = 300 / throughput."
)


async def generate_timed(card):
    started = time.perf_counter()
    extra_body = {
        "structured_outputs": {"json": json_schema},
        "repetition_penalty": CHOSEN["repetition_penalty"],
    }
    if CHOSEN["top_k"] is not None:
        extra_body["top_k"] = CHOSEN["top_k"]
    try:
        await async_client.chat.completions.create(
            model="Qwen/Qwen2.5-1.5B-Instruct",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": json.dumps(card, ensure_ascii=False)},
            ],
            temperature=CHOSEN["temperature"],
            top_p=CHOSEN["top_p"],
            max_tokens=CHOSEN["max_tokens"],
            extra_body=extra_body,
            timeout=180,
        )
        error = None
    except Exception as exc:
        error = str(exc)
    return {
        "product_id": card["product_id"],
        "latency": time.perf_counter() - started,
        "error": error,
    }


def choose_concurrency(rows):
    """Берём уровень, пока throughput растёт. Прирост меньше 10% — плато."""
    usable = [row for row in rows if row["usable"]]
    if not usable:
        return None
    chosen = usable[0]
    for row in usable[1:]:
        gain = (row["throughput"] - chosen["throughput"]) / chosen["throughput"]
        if gain < 0.10:
            break
        chosen = row
    return chosen


print("Прогрев")
await generate_timed(speed_items[0])

speed_table = []
for level in CONCURRENCY_LEVELS:
    print(f"concurrency {level}...", flush=True)
    timed, elapsed = await run_concurrent(speed_items, level, generate_timed)
    latencies = [item["latency"] for item in timed]
    errors = sum(item["error"] is not None for item in timed)
    throughput = len(timed) / elapsed
    latency = sum(latencies) / len(latencies)
    row = {
        "concurrency": level,
        "throughput": throughput,
        "latency": latency,
        "time_300": 300 / throughput,
        "errors": errors,
        "usable": errors <= len(timed) // 5 and throughput > 0,
    }
    speed_table.append(row)
    print(
        f"  throughput {throughput:.2f} карточек/с, "
        f"latency {latency:.2f} с, "
        f"на 300 — {row['time_300']:.0f} с, ошибок {errors}",
        flush=True,
    )

header = f"{'concurrency':>12} {'throughput':>12} {'latency':>10} {'время на 300':>14} {'ошибки':>8}"
print()
print(header)
print("-" * len(header))
for row in speed_table:
    print(
        f"{row['concurrency']:>12} {row['throughput']:>12.2f} {row['latency']:>10.2f} "
        f"{row['time_300']:>12.0f} с {row['errors']:>8}"
    )

chosen_speed = choose_concurrency(speed_table)
if chosen_speed is None:
    print("Ни один уровень не дал устойчивый прогон.")
else:
    chosen_concurrency = chosen_speed["concurrency"]
    print(
        f"\nВыбрано concurrency {chosen_concurrency}: "
        f"throughput {chosen_speed['throughput']:.2f} карточек/с, "
        f"latency {chosen_speed['latency']:.2f} с, "
        f"оценка на 300 — {chosen_speed['time_300']:.0f} с."
    )
    if chosen_speed["time_300"] <= BENCHMARK_LIMIT_SECONDS:
        print("Оценка укладывается в 8 минут. Полный бенчмарк — следующая ячейка.")
    else:
        print(
            "Оценка длиннее 8 минут. "
            "Это последняя точка до плато среди измеренных."
        )
````
